## Problem 2: Create buffers around shopping centers (3 points)

Let’s continue with our case study and calculate a 1.5 km buffer around the geocoded points.

### a) Load the shopping centre data set (1 point)

First, load the shopping centre *point* data set you created in *problem 1* into a `GeoDataFrame` called `shopping_centres`.

```python
import geopandas as gpd
import pandas as pd
from geopandas.tools import geocode

# 1. كتابة البيانات النصية
data = """id;name;addr
1000;Itis;Itäkatu 1-7, 00930 Helsinki, Finland
1001;Forum;Mannerheimintie 20, 00100 Helsinki, Finland
1002;Iso-omena;Piispansilta 11, 02230 Espoo, Finland
1003;Sello;Leppävaarankatu 3-9, 02600 Espoo, Finland
1004;Jumbo;Vantaanportinkatu 3, 01510 Vantaa, Finland
1005;REDI;Hermannin rantatie 5, 00580 Helsinki, Finland
1006;Tripla;Firdonkatu 2, 00520 Helsinki, Finland"""

with open("shopping_centres.txt", "w") as f:
    f.write(data)

# 2. تحويل العناوين وإعادة الإسقاط وحفظ الملف
df = pd.read_csv("shopping_centres.txt", sep=";")
geo = geocode(df["addr"], provider="nominatim", user_agent="autogis_app")
df["geometry"] = geo["geometry"]
gdf = gpd.GeoDataFrame(df, crs="EPSG:4326").to_crs("EPSG:3879")
gdf.to_file("shopping_centres.gpkg")
```

### b) Create a *buffer* around the points (1 points)

Calculate a 1.5 km buffer for each geocoded point. Overwrite the `geometry` column with the new buffer geometry.

Use the [`geopandas.GeoDataFrame.buffer()` method](http://geopandas.org/geometric_manipulations.html#GeoSeries.buffer), that uses shapely’s [`buffer()`](http://toblerity.org/shapely/manual.html#object.buffer) in the background. You only need to care about the `distance` parameter, don’t worry about the possible other arguments.

```python
import pandas as pd
import geopandas as gpd
from geopandas.tools import geocode

# 1. تجهيز البيانات وتحويل نظام الإحداثيات إلى EPSG:3879
data = """id;name;addr
1000;Itis;Itäkatu 1-7, 00930 Helsinki, Finland
1001;Forum;Mannerheimintie 20, 00100 Helsinki, Finland
1002;Iso-omena;Piispansilta 11, 02230 Espoo, Finland
1003;Sello;Leppävaarankatu 3-9, 02600 Espoo, Finland
1004;Jumbo;Vantaanportinkatu 3, 01510 Vantaa, Finland
1005;REDI;Hermannin rantatie 5, 00580 Helsinki, Finland
1006;Tripla;Firdonkatu 2, 00520 Helsinki, Finland"""

with open("shopping_centres.txt", "w") as f:
    f.write(data)

df = pd.read_csv("shopping_centres.txt", sep=";")
geo = geocode(df["addr"], provider="nominatim", user_agent="autogis_app")
df["geometry"] = geo["geometry"]
shopping_centres = gpd.GeoDataFrame(df, crs="EPSG:4326").to_crs("EPSG:3879")

# 2. تطبيق النطاق العازل (Buffer) بمقدار 1500 متر
shopping_centres["geometry"] = shopping_centres.buffer(1500)
```
